```python
import numpy as np

students = np.array(["Asha", "Ravi", "Meena", "Kiran", "Swetha", "Harsh", "Rupesh", "Akhil", "Divya", "Teja", "Nikhil", "Priya"])
marks = np.array([78, 85, 62, 90, 71, 89, 70, 65, 92, 55, 50, 98])
```

```python
mask = marks > 70
print("Boolean array:", mask)
print("Filtered values:", marks[mask])
```

```python
print("Above 70   :", marks[marks > 70])
print("Below 60   :", marks[marks < 60])
print("Equal to 70:", marks[marks == 70])
print("At least 85:", marks[marks >= 85])
```

```python
for threshold in [50, 60, 70, 80, 90]:
    above = marks[marks > threshold]
    print(f"Marks above {threshold}: {above}  (count = {len(above)})")
```

```python
between = marks[(marks >= 60) & (marks < 80)]
extremes = marks[(marks < 55) | (marks > 95)]
print("60 to 79:", between)
print("Below 55 or above 95:", extremes)
```

```python
print("Students with marks >= 85:", students[marks >= 85])
```
